# Desenvolvimento prático da E3 — Sprint 4

Esta seção reúne a execução dos modelos, a demonstração de previsões, a avaliação das métricas e a geração dos artefatos da entrega.

> **Nota metodológica:** os modelos e as métricas disponíveis neste pacote foram produzidos com dados sintéticos. As previsões demonstrativas abaixo usam registros fictícios e não representam contribuintes reais.

```python
# Bibliotecas principais
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import json
import sys
from pathlib import Path

# Caminhos do projeto
PROJETO = Path("/content/Sprint4")
PACOTE = Path("/content/PacoteE3/E3_Notebook_e_Pacote")
MODELOS = PACOTE / "models_sprint4_v1"

# Conferir ambiente
import sklearn

print("Python:", sys.version)
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Scikit-learn:", sklearn.__version__)
print("Ambiente preparado!")
```

## 1. Carregamento dos modelos

Carregamos os modelos M1 (inadimplência) e M2 (recuperabilidade) já serializados no pacote da E3. A versão do scikit-learn deve ser compatível com a versão usada na serialização.

```python
import joblib
import json
from pathlib import Path

# Localização dos modelos
PASTA_MODELOS = Path("/content/PacoteE3/E3_Notebook_e_Pacote/models_sprint4_v1")

# Carregar os dois modelos
modelo_inadimplencia = joblib.load(
    PASTA_MODELOS / "inadimplencia.joblib"
)

modelo_recuperabilidade = joblib.load(
    PASTA_MODELOS / "recuperabilidade.joblib"
)

# Carregar métricas
with open(PASTA_MODELOS / "metrics.json", encoding="utf-8") as arquivo:
    metricas = json.load(arquivo)

print("Modelos carregados!")
print("Modelo M1:", type(modelo_inadimplencia).__name__)
print("Modelo M2:", type(modelo_recuperabilidade).__name__)
```

## 2. Preparação de entradas demonstrativas

Montamos um registro fictício para cada modelo, respeitando os atributos definidos no contrato de entrada (`feature_contract.json`). Esses registros servem apenas para demonstrar a chamada de previsão.

```python
import pandas as pd

# Atributos exigidos pelo modelo M1
dados_m1 = pd.DataFrame([{
    "inadimplente_exercicio_anterior": 1,
    "inadimplencias_anteriores": 2,
    "exercicios_anteriores": 4,
    "debitos_abertos_t0": 2,
    "valor_relativo_anterior": 1.5
}])

# Atributos exigidos pelo modelo M2
dados_m2 = pd.DataFrame([{
    "idade_divida_dias": 180,
    "dias_desde_inscricao": 90,
    "valor_relativo_credito": 1.2,
    "creditos_inscritos_abertos_imovel": 2,
    "inadimplencias_imovel_t0": 1,
    "inadimplente_ultimo_exercicio": 1
}])

print("Dados de entrada preparados!")
display(dados_m1)
display(dados_m2)
```

## 3. Execução das previsões

Aplicamos `predict` e `predict_proba` aos registros demonstrativos. A classe prevista e a probabilidade devem ser interpretadas em conjunto: no M2, por exemplo, a classe positiva pode ser retornada mesmo com probabilidade estimada baixa, devido ao limiar de classificação configurado.

```python
# Previsão de inadimplência (M1)
previsao_m1 = modelo_inadimplencia.predict(dados_m1)
probabilidade_m1 = modelo_inadimplencia.predict_proba(dados_m1)

# Previsão de recuperabilidade (M2)
previsao_m2 = modelo_recuperabilidade.predict(dados_m2)
probabilidade_m2 = modelo_recuperabilidade.predict_proba(dados_m2)

# Exibir resultados
print("=== MODELO M1: INADIMPLÊNCIA ===")
print("Previsão:", "Inadimplente" if previsao_m1[0] == 1 else "Não inadimplente")
print("Probabilidade de inadimplência:", round(probabilidade_m1[0][1] * 100, 2), "%")

print("\n=== MODELO M2: RECUPERABILIDADE ===")
print("Previsão:", "Recuperável" if previsao_m2[0] == 1 else "Baixa recuperabilidade")
print("Probabilidade de recuperação:", round(probabilidade_m2[0][1] * 100, 2), "%")
```

## 4. Registro das previsões

Organizamos as saídas dos dois modelos em uma tabela para documentar o teste demonstrativo.

```python
resultados = pd.DataFrame({
    "Modelo": ["M1 - Inadimplência", "M2 - Recuperabilidade"],
    "Previsão": [
        "Inadimplente" if previsao_m1[0] == 1 else "Não inadimplente",
        "Recuperável" if previsao_m2[0] == 1 else "Baixa recuperabilidade"
    ],
    "Probabilidade (%)": [
        round(probabilidade_m1[0][1] * 100, 2),
        round(probabilidade_m2[0][1] * 100, 2)
    ]
})

display(resultados)
```

## 5. Avaliação dos modelos

Apresentamos as métricas calculadas na avaliação do projeto. Elas se referem ao conjunto de teste sintético e não devem ser interpretadas como desempenho em produção.

```python
# Exibir métricas de avaliação do projeto

for modelo, dados in metricas.items():
    print(f"\n===== {modelo} =====")

    for metrica, valor in dados.items():
        if isinstance(valor, (int, float)):
            print(f"{metrica}: {valor:.4f}")
        else:
            print(f"{metrica}: {valor}")
```

## 6. Comparação das métricas

A tabela resume acurácia, precisão, recall, F1-score e ROC-AUC para M1 e M2.

```python
avaliacao_modelos = pd.DataFrame([
    {
        "Modelo": "M1 - Inadimplência",
        "Acurácia": 0.7111,
        "Precisão": 0.5994,
        "Recall": 0.6425,
        "F1-score": 0.6202,
        "ROC-AUC": 0.7406
    },
    {
        "Modelo": "M2 - Recuperabilidade",
        "Acurácia": 0.1240,
        "Precisão": 0.1186,
        "Recall": 0.9938,
        "F1-score": 0.2119,
        "ROC-AUC": 0.5079
    }
])

display(avaliacao_modelos.round(4))
```

## 7. Visualização comparativa

O gráfico facilita a comparação entre as métricas dos dois modelos.

```python
metricas_grafico = avaliacao_modelos.set_index("Modelo")[
    ["Acurácia", "Precisão", "Recall", "F1-score", "ROC-AUC"]
]

metricas_grafico.plot(
    kind="bar",
    figsize=(12, 6),
    ylim=(0, 1),
    rot=0
)

plt.title("Comparação de desempenho dos modelos M1 e M2")
plt.ylabel("Resultado")
plt.xlabel("Modelos")
plt.legend(title="Métricas")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 8. Relatório de avaliação

Geramos um relatório em Markdown com os resultados, a interpretação e as limitações dos modelos.

```python
from pathlib import Path

PASTA_RESULTADOS = Path("/content/PacoteE3/E3_Notebook_e_Pacote/resultados")
PASTA_RESULTADOS.mkdir(parents=True, exist_ok=True)

relatorio = f"""
# Relatório de Avaliação — Sprint 4 / E3

## 1. Objetivo
Avaliar os modelos de inteligência artificial desenvolvidos
para inadimplência e recuperabilidade de créditos de IPTU.

## 2. Modelo M1 — Inadimplência
- Acurácia: 71,11%
- Precisão: 59,94%
- Recall: 64,25%
- F1-score: 62,02%
- ROC-AUC: 0,7406

## 3. Modelo M2 — Recuperabilidade
- Acurácia: 12,40%
- Precisão: 11,86%
- Recall: 99,38%
- F1-score: 21,19%
- ROC-AUC: 0,5079

## 4. Análise dos resultados
O modelo M1 apresentou capacidade de discriminação
nos dados sintéticos utilizados.

O modelo M2 apresentou baixa capacidade de discriminação,
com ROC-AUC próximo de 0,5 e alta taxa de previsões positivas.

## 5. Limitações
Os dados utilizados são sintéticos. Portanto, os resultados
não comprovam desempenho sobre contribuintes reais de Palmas.

## 6. Conclusão
O M1 apresenta resultados mais promissores para a tarefa
de classificação de inadimplência. O M2 necessita de
aprimoramento e de dados com maior capacidade informativa.
"""

arquivo_relatorio = PASTA_RESULTADOS / "relatorio_avaliacao_E3.md"
arquivo_relatorio.write_text(relatorio, encoding="utf-8")

print("Relatório criado com sucesso!")
print("Local:", arquivo_relatorio)
```

## 9. Exportação dos resultados

Exportamos a tabela comparativa em CSV e as métricas completas em JSON.

```python
# Salvar tabela comparativa em CSV
avaliacao_modelos.to_csv(
    PASTA_RESULTADOS / "avaliacao_modelos_E3.csv",
    index=False,
    encoding="utf-8-sig"
)

# Salvar métricas completas em JSON
with open(
    PASTA_RESULTADOS / "metricas_completas_E3.json",
    "w",
    encoding="utf-8"
) as arquivo:
    json.dump(metricas, arquivo, ensure_ascii=False, indent=4)

print("Arquivos exportados:")
for arquivo in PASTA_RESULTADOS.iterdir():
    print("-", arquivo.name)
```

## 10. Empacotamento dos artefatos

A última célula reúne modelos, resultados e notebook em um arquivo ZIP. **Atenção:** para que o ZIP contenha esta versão atualizada do notebook, substitua o notebook original dentro da pasta de entrega antes de gerar o ZIP final.
